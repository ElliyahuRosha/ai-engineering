```python
# Training data
xs = [1, 2, 3, 4, 5, 6, 7, 8]
ys = [2.9, 3.4, 4.9, 4.7, 6.2, 6.9, 7.3, 8.6]

# Parameters and learning rate
w, b = 0.0, 0.0
lr = 0.01
```

```python
for epoch in range(1000):
    # Reset gradient accumulators
    dw, db = 0.0, 0.0

    # BATCH GD: compute gradients over ALL samples
    for x, y in zip(xs, ys):
        # Forward pass
        pred = w * x + b
        err = pred - y

        # Accumulate MSE gradients (no updates yet)
        dw += 2/len(xs) * err * x
        db += 2 * err / len(xs)

    # BATCH GD: update ONLY after processing all samples
    w -= lr * dw
    b -= lr * db
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Modern, clean visualization
fig, ax = plt.subplots(figsize=(10, 6))
fig.patch.set_facecolor("#f8fafc")
ax.set_facecolor("#f8fafc")

# Training data
ax.scatter(xs, ys, s=95, color="#2563eb",
           edgecolors="white", linewidths=1.8,
           zorder=3, label="Training data")

# Learned regression line
x_line = np.linspace(min(xs) - 0.5, max(xs) + 0.5, 200)
y_line = w * x_line + b

ax.plot(x_line, y_line, color="#f97316",
        linewidth=2.8, zorder=2,
        label=f"Prediction: y = {w:.3f}x + {b:.3f}")

# Final training error
mse = np.mean((w * np.array(xs) + b - np.array(ys)) ** 2)

# Labels and styling
ax.set_title("Linear Regression | Batch Gradient Descent",
             fontsize=17, fontweight="bold", pad=22)
ax.set_xlabel("X", fontsize=12, labelpad=10)
ax.set_ylabel("Y", fontsize=12, labelpad=10)

ax.grid(True, linestyle="--", alpha=0.18)
ax.spines[["top", "right", "left"]].set_visible(False)
ax.tick_params(length=0, labelsize=11)
ax.legend(frameon=False, fontsize=11, loc="upper left")

# Model performance
ax.text(0.97, 0.06, f"MSE = {mse:.4f}",
        transform=ax.transAxes, ha="right",
        fontsize=11, fontweight="bold", color="#475569",
        bbox=dict(boxstyle="round,pad=0.6",
                  facecolor="#e2e8f0", edgecolor="none"))

plt.tight_layout()
plt.show()
```
